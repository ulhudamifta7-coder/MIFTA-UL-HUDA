```python
import random

# Objective function (we want to maximize this)
def f(x):
    return -x**2 + 4*x

# Hill Climbing Algorithm
def hill_climb(start):
    current = start
    current_value = f(current)

    step_size = 0.1  # small movement step

    while True:
        # generate neighbors (left and right)
        left = current - step_size
        right = current + step_size

        left_value = f(left)
        right_value = f(right)

        # choose best neighbor
        if left_value > current_value:
            current = left
            current_value = left_value
        elif right_value > current_value:
            current = right
            current_value = right_value
        else:
            break  # no improvement → stop

    return current, current_value


# starting point (random or fixed)
start = random.uniform(-10, 10)

best_x, best_val = hill_climb(start)

print("Starting point:", start)
print("Best solution x:", best_x)
print("Best value f(x):", best_val)
```
